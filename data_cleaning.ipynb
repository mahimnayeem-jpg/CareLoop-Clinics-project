```python
import pandas as pd
import numpy as np

# Convert age to numeric

appointments_clean["age"] = pd.to_numeric(
    appointments_clean["age"],
    errors="coerce"
)

#Identify invalid ages
invalid_age = (
    (appointments_clean["age"] < 1)|
    (appointments_clean["age"] >100)
)

print("Invalid ages:", invalid_age.sum())

# Replace invalid ages with missing values
appointments_clean.loc[invalid_age, "age"] = np.nan

#Fill missing ages using median
age_median = appointments_clean["age"].median()

appointments_clean["age"] = (appointments_clean["age"].fillna(age_median))

print("Median age:", age_median)
print("Missing ages remaining:", appointments_clean["age"].isnull().sum())

```

```python
# Convert distance to numeric
appointments_clean["distance_km"] = pd.to_numeric(
  appointments_clean["distance_km"],
    errors="coerce"
)

# Negative distance is invalid
appointments_clean.loc[
    appointments_clean["distance_km"] < 0, "distance_km"
    ] = np.nan

# Fill missing values with median
distance_median = appointments_clean["distance_km"].median()

appointments_clean["distance_km"] = (
    appointments_clean["distance_km"].fillna(distance_median)
 )

print("Distance median:", distance_median)
print("Missing distance values:",
      appointments_clean["distance_km"].isnull().sum())




```

```python
# Convert free into numeric
appointments_clean["fee_inr"] = pd.to_numeric(
    appointments_clean["fee_inr"],
    errors="coerce"
)

# Replace negative fees with missing values
appointments_clean.loc[
    appointments_clean["fee_inr"] < 0,
    "fee_inr"
] = np.nan

# Fill missing fees with median
fee_median = appointments_clean["fee_inr"].median()

appointments_clean["fee_inr"] = (
    appointments_clean["fee_inr"].fillna(fee_median)
)

# claculate IQR
Q1 = appointments_clean["fee_inr"].quantile(0.25)
Q3 = appointments_clean["fee_inr"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Fee lower limit:", lower_limit)
print("Fee upper limit:", upper_limit)

#cap outliers
appointments_clean["fee_inr"] = (
    appointments_clean["fee_inr"].clip(
        lower=lower_limit,
        upper=upper_limit
    )
)

print("Fee outliers handled")
```

```python

import pandas as pd
import numpy as np
from datetime import date, datetime

def sql_value(value):

    if pd.isna(value):
        return "NULL"

    if isinstance(value, (pd.Timestamp, np.datetime64)):
        return "'" + pd.Timestamp(value).strftime("%Y-%m-%d") + "'"

    if isinstance(value, (datetime, date)):
        return "'" + value.strftime("%Y-%m-%d") + "'"

    if isinstance(value, (int, np.integer)):
        return str(int(value))

    if isinstance(value, (float, np.floating)):
        return str(float(value))

    text_value = str(value).replace("'", "''")
    return "'" + text_value + "'"


def create_sql_table(df, table_name):

    definitions = []

    for column in df.columns:

        dtype = df[column].dtype

        if pd.api.types.is_datetime64_any_dtype(dtype):
            sql_type = "DATE"

        elif pd.api.types.is_integer_dtype(dtype):
            sql_type = "BIGINT"

        elif pd.api.types.is_float_dtype(dtype):
            sql_type = "DOUBLE"

        else:
            sql_type = "TEXT"

        definitions.append(
            f"`{column}` {sql_type}"
        )

    sql = f"DROP TABLE IF EXISTS `{table_name}`;\n"
    sql += f"CREATE TABLE `{table_name}` (\n"
    sql += ",\n".join(definitions)
    sql += "\n);\n\n"

    columns = ", ".join(
        f"`{column}`" for column in df.columns
    )

    for row in df.itertuples(index=False, name=None):

        values = ", ".join(
            sql_value(value) for value in row
        )

        sql += (
            f"INSERT INTO `{table_name}` "
            f"({columns}) VALUES ({values});\n"
        )

    return sql


# Create SQL script
sql_script = """
CREATE DATABASE IF NOT EXISTS careloop_clinics;
USE careloop_clinics;

"""

sql_script += create_sql_table(
    clinics_clean, "clinics"
)

sql_script += create_sql_table(
    appointments_clean, "appointments"
)

# Save SQL file
with open("load_data.sql", "w", encoding="utf-8") as f:
    f.write(sql_script)

print("load_data.sql created successfully!")
```

```python

# Select the target column
y = appointments["no_show"]

# Select input features
feature_columns = [
    "age",
    "gender",
    "distance_km",
    "fee_inr",
    "booking_channel",
    "visit_type",
    "reminder_sent",
    "prior_no_shows",
    "lead_days"
]

# Keep only columns available in the dataset
feature_columns = [
    col for col in feature_columns
    if col in appointments.columns
]

X = appointments[feature_columns].copy()

print("Input features:")
print(X.columns.tolist())

print("\nTarget variable:")
print(y.name)

print("\nInput data shape:", X.shape)
print("Target data shape:", y.shape)
```

```python

X = pd.get_dummies(X, drop_first=True)

print("Data converted successfully!")
print("Final input shape:", X.shape)

display(X.head())
```

```python

from sklearn.model_selection import train_test_split

# Split the dataset
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Data splitting completed!")

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)

print("Training target:", y_train.shape)
print("Testing target:", y_test.shape)
```

```python

from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score, classification_report

# Create Logistic Regression model
log_model = Pipeline([
    ("scaler", StandardScaler()),
    ("model", LogisticRegression(max_iter=1000))
])

# Train the model
log_model.fit(X_train, y_train)

# Predict test data
y_pred_log = log_model.predict(X_test)

# Check accuracy
accuracy = accuracy_score(y_test, y_pred_log)

print("LOGISTIC REGRESSION RESULTS")
print("Accuracy:", round(accuracy * 100, 2), "%")

print("\nClassification Report:")
print(classification_report(y_test, y_pred_log))
```

```python

from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report

# Create Decision Tree model
dt_model = DecisionTreeClassifier(
    max_depth=5,
    random_state=42
)

# Train the model
dt_model.fit(X_train, y_train)

# Predict test data
y_pred_dt = dt_model.predict(X_test)

# Calculate accuracy
dt_accuracy = accuracy_score(y_test, y_pred_dt)

print("DECISION TREE RESULTS")
print("Accuracy:", round(dt_accuracy * 100, 2), "%")

print("\nClassification Report:")
print(classification_report(y_test, y_pred_dt))
```

```python

from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report

# Create Random Forest model
rf_model = RandomForestClassifier(
    n_estimators=100,
    max_depth=8,
    random_state=42,
    class_weight="balanced"
)

# Train the model
rf_model.fit(X_train, y_train)

# Predict test data
y_pred_rf = rf_model.predict(X_test)

# Calculate accuracy
rf_accuracy = accuracy_score(y_test, y_pred_rf)

print("RANDOM FOREST RESULTS")
print("Accuracy:", round(rf_accuracy * 100, 2), "%")

print("\nClassification Report:")
print(classification_report(y_test, y_pred_rf))
```

```python

import pandas as pd
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    roc_auc_score
)
from sklearn.model_selection import StratifiedKFold, cross_val_score

# Store all three models and predictions
models = {
    "Logistic Regression": (log_model, y_pred_log),
    "Decision Tree": (dt_model, y_pred_dt),
    "Random Forest": (rf_model, y_pred_rf)
}

results = []

# Compare model performance
for name, (model, predictions) in models.items():

    accuracy = accuracy_score(y_test, predictions)
    precision = precision_score(y_test, predictions, zero_division=0)
    recall = recall_score(y_test, predictions, zero_division=0)
    auc = roc_auc_score(y_test, model.predict_proba(X_test)[:, 1])

    results.append({
        "Model": name,
        "Accuracy": round(accuracy, 3),
        "Precision": round(precision, 3),
        "Recall": round(recall, 3),
        "ROC_AUC": round(auc, 3)
    })

# Display comparison
comparison = pd.DataFrame(results)

print("MODEL PERFORMANCE COMPARISON")
display(comparison)
```

```python

from sklearn.model_selection import StratifiedKFold, cross_val_score

# Create 5-fold cross-validation
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

models_to_test = {
    "Logistic Regression": log_model,
    "Decision Tree": dt_model,
    "Random Forest": rf_model
}

cv_results = []

for name, model in models_to_test.items():

    scores = cross_val_score(
        model,
        X_train,
        y_train,
        cv=cv,
        scoring="roc_auc"
    )

    cv_results.append({
        "Model": name,
        "Mean ROC-AUC": round(scores.mean(), 3),
        "Standard Deviation": round(scores.std(), 3)
    })

cv_table = pd.DataFrame(cv_results)

print("CROSS-VALIDATION RESULTS")
display(cv_table)
```

```python

# Select the model with the highest mean ROC-AUC
best_model_name = cv_table.loc[
    cv_table["Mean ROC-AUC"].idxmax(), "Model"
]

best_model = models_to_test[best_model_name]

print("BEST MODEL SELECTED:", best_model_name)
print(
    "Cross-validation ROC-AUC:",
    cv_table["Mean ROC-AUC"].max()
)

# Train the selected model
best_model.fit(X_train, y_train)

print("Best model trained successfully!")
```

```python

# Predict no-show status
final_predictions = best_model.predict(X_test)

# Get no-show probabilities
no_show_probability = best_model.predict_proba(X_test)[:, 1]

# Create prediction results
prediction_results = pd.DataFrame({
    "Actual_No_Show": y_test.values,
    "Predicted_No_Show": final_predictions,
    "No_Show_Probability": no_show_probability
})

# Display results
display(prediction_results.head(10))

# Count predicted no-shows
print(
    "Predicted no-show patients:",
    (final_predictions == 1).sum()
)
```

```python

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)
import joblib

# Calculate final evaluation metrics
print("FINAL MODEL:", best_model_name)
print("Accuracy:", round(accuracy_score(y_test, final_predictions), 3))
print("Precision:", round(precision_score(y_test, final_predictions, zero_division=0), 3))
print("Recall:", round(recall_score(y_test, final_predictions, zero_division=0), 3))
print("F1 Score:", round(f1_score(y_test, final_predictions, zero_division=0), 3))
print("ROC-AUC:", round(roc_auc_score(y_test, no_show_probability), 3))

# Save prediction results
prediction_results.to_csv(
    "patient_no_show_predictions.csv",
    index=False
)

# Save trained model
joblib.dump(best_model, "best_no_show_model.pkl")

print("\nPrediction file saved successfully!")
print("Model saved successfully!")
```

```python

import pandas as pd
import time
import json
from google import genai
from google.genai import types

# Use the Gemini Flash Lite model
MODEL_NAME = "gemini-3.5-flash-lite"

results = []

for index, row in notes_df.iterrows():

    note = str(row["note_text"])

    prompt = f"""
    Analyse this patient follow-up note.

    Identify:
    1. Main reason for missing the appointment
    2. Short explanation
    3. Recommended clinic follow-up action

    Patient note: {note}

    Return JSON with these fields:
    reason_category, reason_summary, recommended_action
    """

    try:
        response = client.models.generate_content(
            model=MODEL_NAME,
            contents=prompt,
            config=types.GenerateContentConfig(
                response_mime_type="application/json"
            )
        )

        analysis = json.loads(response.text)

        results.append({
            "note_id": row["note_id"],
            "appointment_id": row["appointment_id"],
            "reason_category": analysis["reason_category"],
            "reason_summary": analysis["reason_summary"],
            "recommended_action": analysis["recommended_action"]
        })

        print("Successfully analysed:", row["note_id"])

    except Exception as e:
        print("Error:", row["note_id"], str(e)[:200])

    # Wait 5 seconds between requests
    time.sleep(5)

# Save results
results_df = pd.DataFrame(results)
results_df.to_csv("notes_analysed.csv", index=False)

print("Analysis completed!")
print("Successfully analysed:", len(results_df))
print("File saved: notes_analysed.csv")
```

```python

import pandas as pd
import json
import time
import os

# Load the 19 already analysed notes
if os.path.exists("notes_analysed.csv"):
    old_results = pd.read_csv("notes_analysed.csv")
else:
    old_results = pd.DataFrame()

done_ids = set(old_results["note_id"].astype(str))

new_results = []

for index, row in notes_df.iterrows():

    note_id = str(row["note_id"])

    # Skip notes already analysed
    if note_id in done_ids:
        continue

    note = str(row["note_text"])
    success = False

    for attempt in range(3):
        try:
            prompt = f"""
            Analyse this patient follow-up note:
            {note}

            Return JSON with:
            reason_category,
            reason_summary,
            recommended_action
            """

            response = client.models.generate_content(
                model="gemini-3.5-flash-lite",
                contents=prompt,
                config={
                    "response_mime_type": "application/json"
                }
            )

            analysis = json.loads(response.text)

            new_results.append({
                "note_id": note_id,
                "appointment_id": row["appointment_id"],
                "reason_category": analysis["reason_category"],
                "reason_summary": analysis["reason_summary"],
                "recommended_action": analysis["recommended_action"]
            })

            print("Successfully analysed:", note_id)
            success = True
            break

        except Exception as e:
            print("Retry", attempt + 1, "for", note_id, ":", str(e)[:100])
            time.sleep(15 * (attempt + 1))

    if not success:
        print("Still pending:", note_id)

    time.sleep(5)

# Combine old and new results
final_results = pd.concat(
    [old_results, pd.DataFrame(new_results)],
    ignore_index=True
)

final_results = final_results.drop_duplicates(
    subset="note_id",
    keep="last"
)

final_results.to_csv("notes_analysed.csv", index=False)

print("Total analysed:", len(final_results))
print("Remaining:", 30 - len(final_results))
print("Saved notes_analysed.csv")
```
